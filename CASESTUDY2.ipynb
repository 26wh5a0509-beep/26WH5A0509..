```
import pandas as pd

# 7. Create a DataFrame from the dataset
data = {
    "Student": ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8", "S9", "S10"],
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 4, 6, 9],
    "Attendance": [65, 70, 75, 80, 85, 90, 92, 78, 88, 95],
    "Assignment_Score": [55, 60, 65, 72, 78, 85, 88, 70, 80, 92],
    "Previous_Score": [58, 62, 68, 70, 76, 82, 86, 72, 79, 90],
    "Final_Score": [60, 64, 70, 75, 82, 88, 91, 74, 84, 94]
}
df = pd.DataFrame(data)

# 8. Display the complete dataset
print("Complete Dataset:")
print(df)

# 9. Calculate the average of numerical columns
print("\nAverage values:")
print(df[[
    "Study_Hours",
    "Attendance",
    "Assignment_Score",
    "Previous_Score",
    "Final_Score"
]].mean())

# 10. Find the student with the highest Final Score
highest = df.loc[df["Final_Score"].idxmax()]

print("\nStudent with the highest Final Score:")
print(highest)

# 11. Find the student with the lowest Final Score
lowest = df.loc[df["Final_Score"].idxmin()]

print("\nStudent with the lowest Final Score:")
print(lowest)

# 12. Display students whose Attendance is greater than 80%
attendance_above_80 = df[df["Attendance"] > 80]

print("\nStudents with Attendance greater than 80%:")
print(attendance_above_80)

# 13. Display students whose Final Score is greater than 80
final_above_80 = df[df["Final_Score"] > 80]

print("\nStudents with Final Score greater than 80:")
print(final_above_80)

# 14. Statistical summary using describe()
print("\nStatistical Summary:")
print(df.describe())
```

```
import numpy as np

# Extract Final_Score as a NumPy array
final_scores = np.array(df["Final_Score"])

# Calculate statistical measures
mean_score = np.mean(final_scores)
median_score = np.median(final_scores)
maximum_score = np.max(final_scores)
minimum_score = np.min(final_scores)
standard_deviation = np.std(final_scores)

print("Final Scores:", final_scores)
print("Mean:", mean_score)
print("Median:", median_score)
print("Maximum:", maximum_score)
print("Minimum:", minimum_score)
print("Standard Deviation:", standard_deviation)
```

```
import numpy as np

def L(x):
    return 5*x - 0.2*x**2

# Generate points between 0 and 5
x = np.linspace(0, 5, 1000)
y = L(x)
# Numerical integration using trapezoidal rule
area = np.trapezoid(y, x)
print("Numerical approximation:", area)
```

```
import numpy as np

# Input variables
X = df[["Study_Hours", "Attendance", "Assignment_Score"]].values

# Target variable
Y = df["Final_Score"].values

# Add a column of 1s for the intercept
X = np.column_stack((np.ones(len(X)), X))

# Calculate coefficients
beta = np.linalg.inv(X.T @ X) @ X.T @ Y
print("Coefficients:")
print(beta)
print("\nIntercept (b0):", beta[0])
print("Study Hours coefficient (b1):", beta[1])
print("Attendance coefficient (b2):", beta[2])
print("Assignment Score coefficient (b3):", beta[3])
```
