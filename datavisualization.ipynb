```
import seaborn as sns
tips = sns.load_dataset("tips")
tips.head()
```

```
import matplotlib.pyplot as plt
counts = tips["smoker"].value_counts()
plt.pie(
    counts,
    labels=counts.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Smoker Distribution")
plt.show()
```

```
plt.hist(
    tips["total_bill"],
    bins=20,
    edgecolor="black"
)
plt.title("Distribution of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

```
sns.scatterplot(
data=tips,
x="total_bill",
y="tip",
hue="smoker"
)
```

```
sns.barplot(
data=tips,
x="day",
y="total_bill",
hue="sex"

)
```

```
sns.boxplot(
data=tips,
x="sex",
y="total_bill"
)
```

```
sns.countplot(
data=tips,
x="day"
)
```

```
sns.swarmplot(
data=tips,
x="smoker",
y="tip"
)
```

```
plt.hist(tips["total_bill"], bins=20, edgecolor="black")
plt.title("Distribution of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.grid()
plt.tight_layout()
plt.show()
```

```
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr"]
sales = [20000, 25000, 30000, 35000]
plt.plot(months, sales)
plt.title("Monthly Sales")
plt.xlabel("Month"); plt.ylabel("Sales")
plt.show()
```

```
import matplotlib.pyplot as plt
months = ["January", "February", "March", "April"]
revenue = [20000, 25000, 30000, 35000]
plt.plot(months, revenue, marker="o")
plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue (₹)")
plt.grid()
plt.tight_layout()
plt.show()
```

```
import matplotlib.pyplot as plt
products = ["A", "B", "C", "D"]
sales = [50000, 70000, 40000, 90000]
plt.bar(products, sales)
plt.title("Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales (₹)")
plt.grid(axis="y")
plt.tight_layout()
plt.show()
```

```
import matplotlib.pyplot as plt
methods = ["UPI", "Credit Card", "Debit Card", "Cash"]
percentages = [45, 25, 20, 10]
plt.pie(
    percentages,
    labels=methods,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Customer Payment Methods")
plt.tight_layout()
plt.show()
```
