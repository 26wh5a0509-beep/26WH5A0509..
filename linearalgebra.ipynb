```python
import numpy as np
a = np.array([1, 2, 3])
b = np.array([4, 0, 1])
print(np.dot(a, b))
print(np.cross(a, b))
print(np.linalg.norm(a))
print(np.linalg.norm(a, 1))
cos = (a @ b) / (np.linalg.norm(a) * np.linalg.norm(b))
print(cos)
```

```python
import numpy as np
A = np.array([[2., 1.],
              [1., 3.]])
B = np.array([[1., 2.],
              [3., 4.]])
Q = np.array([[0., -1.],
              [1.,  0.]])
print(A @ B)
print(A.T)
print(np.linalg.inv(A))
print(np.trace(A))
print(np.allclose(A, A.T))
print(np.allclose(Q.T @ Q, np.eye(2)))
```

```python
import numpy as np
theta = np.radians(25)
R = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta),  np.cos(theta)]
])
point = np.array([1., 0.])
rotated_point = R @ point
print(R)
print(rotated_point)
```

```python
import numpy as np
A = np.array([[2., 0.],
              [0., 3.]])
vals, vecs = np.linalg.eig(A)
print(vals)       # eigenvalues
print(vecs)       # eigenvectors
print(np.allclose(A @ vecs[:, 0], vals[0] * vecs[:, 0]))
```

```python
import numpy as np
A = np.array([
    [2, 1, -1],
    [-3, -1, 2],
    [-2, 1, 2]
])
b = np.array([8, -11, -3])
x = np.linalg.solve(A, b)
print(x)
print(np.linalg.matrix_rank(A))
```

```python
A = np.array([
    [2, 1],
    [1, 3]
])
A_inverse = np.linalg.inv(A)
print("Matrix A:")
print(A)
print("\nInverse of A:")
print(A_inverse)
```

```python
A = np.array([
    [2, 1],
    [1, 3]
])
det = np.linalg.det(A)
print("Determinant:", det)
```

```python
A = np.array([
    [2, 1],
    [1, 2]
])
eigenvalues, eigenvectors = np.linalg.eig(A)
print("Eigenvalues:")
print(eigenvalues)
print("\nEigenvectors:")
print(eigenvectors)
```
