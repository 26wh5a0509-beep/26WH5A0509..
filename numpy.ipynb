```python
import numpy as np
scores = np.array([88, 92, 79, 95, 61, 74, 88])
print('Mean :', scores.mean())
print('Median :', np.median(scores))
print('Std Dev:', scores.std())
print('Max :', scores.max(), 'at index', scores.argmax())
passed = scores[scores >= 75]
print('Passed :', passed)
print('Pass %%:', 100 * len(passed) / len(scores))
```
